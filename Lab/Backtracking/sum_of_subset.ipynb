# <span style="color:#1f4e79;">Sum of Subsets Problem — Backtracking</span>

The **Sum of Subsets Problem** is a classic backtracking problem where the goal is to find all possible subsets of a given set of numbers whose sum is equal to a target value **S**. Unlike the Dynamic Programming version which often only checks for existence, the Backtracking approach is typically used to explore and generate the actual subsets.

Where:  
- `n = number of elements`  
- `arr[] = set of positive integers`  
- `S = target sum`  
- **Method:** Each element has two recursive choices: **Include (Take)** or **Exclude (Not Take)**.

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`arr = [3, 4, 5, 2], S = 7`  

Output:  
`{3, 4}, {5, 2}`  

Explanation:  
1. $3 + 4 = 7$ (Valid subset)  
2. $5 + 2 = 7$ (Valid subset)  

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`arr = [2, 4, 6], S = 5`  

Output:  
`No valid subset`  

Explanation:  
No combination of the numbers 2, 4, and 6 can be added together to equal exactly 5.

## <span style="color:#1f4e79;">Related Problems</span>

* **Subset Sum (DP):** For existence check only.
* **Combination Sum:** Where elements can be reused.
* **Equal Sum Partition:** Dividing a set into two subsets of equal sum.
* **0/1 Knapsack:** Finding items with max value within a weight limit.
* **Power Set:** Generating all possible $2^n$ subsets.

---

## <span style="color:#1f4e79;">Use Cases</span>

- **Decision Making Systems:** Evaluating all feasible combinations that satisfy a specific constraint.
- **Resource Allocation:** Matching exact budgets or capacity limits with available resources.
- **Scheduling:** Selecting a group of tasks that fit into a fixed total time slot.
- **Cryptography:** Used in subset-based key selection and public-key cryptosystems.
- **Optimization Problems:** Exploring all feasible solution paths when heuristics are not available.
- **Backtracking Practice:** Serving as the introductory problem for "Include/Exclude" recursive patterns.
