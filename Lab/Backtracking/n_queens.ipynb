# <span style="color:#1f4e79;">N-Queens Problem — Backtracking</span>

The **N-Queens Problem** is a classic puzzle and a hallmark of **Backtracking** algorithms. The objective is to place **N queens** on an **N × N chessboard** such that no two queens attack each other. In chess, a queen can attack horizontally, vertically, or diagonally. Therefore, a valid solution requires that no two queens share the same row, column, or diagonal.

Where:  
- `N = dimensions of the board (N × N)`  
- `N = number of queens to be placed`  
- `Output = All valid board configurations or the total count of solutions`

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`N = 4`  

Output:  
`2`  

Explanation:  
There are exactly 2 distinct ways to place 4 queens on a 4x4 board.  
One solution: `[ . Q . . ]`, `[ . . . Q ]`, `[ Q . . . ]`, `[ . . Q . ]`  
Second solution is a mirror image.



---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`N = 2`  

Output:  
`0`  

Explanation:  
On a 2x2 board, it is impossible to place 2 queens such that they do not attack each other (they will always share a row, column, or diagonal).
