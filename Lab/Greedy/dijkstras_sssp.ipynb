# <span style="color:#1f4e79;">Dijkstra’s Single Source Shortest Path Algorithm — Greedy Algorithm</span>

**Dijkstra’s Algorithm** is a classic greedy graph algorithm used to find the shortest path from a single source vertex to all other vertices in a weighted graph. It works by maintaining a set of explored vertices and greedily expanding to the closest unvisited vertex. Note that this algorithm requires **non-negative edge weights** to guarantee correctness.

Where:  
- `V = number of vertices`  
- `E = number of edges`  
- `G(V, E) = weighted graph`  
- `src = source vertex`  
- **Condition:** All edge weights must be $\ge 0$.

---

## <span style="color:#1f4e79;">Example 1:</span>

Input (Graph):  
`Vertices: 5, Source: 0`  
`Edges: 0→1(10), 0→4(5), 1→2(1), 1→4(2), 2→3(4), 3→0(7), 4→1(3), 4→2(9), 4→3(2)`  

Output:  
`Shortest distances from source 0: [0, 8, 9, 7, 5]`  

Explanation:  
- Path to 4: `0 → 4` (dist 5)  
- Path to 1: `0 → 4 → 1` (dist 5+3=8)  
- Path to 3: `0 → 4 → 3` (dist 5+2=7)  
- Path to 2: `0 → 4 → 1 → 2` (dist 5+3+1=9)



---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`Vertices: 3, Source: 0`  
`Edges: 0→1(4), 0→2(1), 2→1(2)`  

Output:  
`Shortest distances: [0, 3, 1]`  

Explanation:  
Direct path $0 \to 1$ is 4, but the path $0 \to 2 \to 1$ has a total weight of $1 + 2 = 3$, which is shorter.

## <span style="color:#1f4e79;">Related Problems</span>

* **Bellman–Ford Algorithm:** Handles negative edge weights.
* **Floyd–Warshall Algorithm:** Finds shortest paths between all pairs of vertices.
* **A* Search Algorithm:** Uses heuristics to find the shortest path faster.
* **Shortest Path in DAG:** Optimized $O(V+E)$ version using Topological Sort.

## <span style="color:#1f4e79;">Use Cases</span>

- **GPS & Navigation:** Calculating the fastest driving route between two points on a map.
- **Network Routing:** OSPF (Open Shortest Path First) protocol uses Dijkstra for routing data packets.
- **Telecommunications:** Optimizing the transmission paths for data to minimize latency.
- **Game Development:** Pathfinding for NPCs (Non-Player Characters) in complex environments.
- **AI & Robotics:** Planning a collision-free path for a robot from point A to point B.
- **Social Networks:** Finding "degrees of separation" or suggested connections between users.
