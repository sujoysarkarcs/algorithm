# <span style="color:#1f4e79;">Kruskal’s Minimum Spanning Tree (MST) — Greedy Algorithm</span>

**Kruskal’s Algorithm** is a classic greedy graph algorithm used to find the **Minimum Spanning Tree (MST)** of a connected, weighted, and undirected graph. Unlike Prim's algorithm which grows from a vertex, Kruskal’s grows the MST by focusing on edges, always selecting the smallest available edge that does not form a cycle.

Where:  
- `V = number of vertices`  
- `E = number of edges`  
- `G(V, E) = weighted, undirected graph`  
- **Result:** The MST contains exactly **V − 1** edges and uses **Disjoint Set Union (DSU)** for efficiency.

---

## <span style="color:#1f4e79;">Example 1:</span>

Input (Graph):  
`Vertices: 4`  
`Edges: (0-1, 10), (0-2, 6), (0-3, 5), (1-3, 15), (2-3, 4)`  

Output:  
`Minimum cost of MST = 19`  

Explanation:  
1. Sort edges: (2-3, 4), (0-3, 5), (0-2, 6), (0-1, 10), (1-3, 15).  
2. Pick (2-3, 4) - No cycle.  
3. Pick (0-3, 5) - No cycle.  
4. Pick (0-2, 6) - **Cycle detected** (0-3-2), discard.  
5. Pick (0-1, 10) - No cycle.  
Total weight: $4 + 5 + 10 = 19$.



---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`Vertices: 3`  
`Edges: (0-1, 1), (1-2, 2), (0-2, 3)`  

Output:  
`Minimum cost of MST = 3`  

Explanation:  
Edges sorted by weight: (0-1, 1), (1-2, 2), (0-2, 3).  
We pick (0-1) and (1-2). Adding (0-2) would form a cycle.  
Total weight: $1 + 2 = 3$.

Kruskal's relies on sorting edges and managing connected components using the **Union-Find** data structure.



1. **Sort:** Arrange all $E$ edges in non-decreasing order of their weight.
2. **Initialize DSU:** Create $V$ subsets, each containing one vertex.
3. **Iterate:** For each sorted edge $(u, v)$ with weight $w$:
   - Find the root of $u$ and $v$ using the `Find` operation.
   - If `Find(u) != Find(v)`, the edge does not form a cycle.
   - Add the edge to MST and perform a `Union` of the two subsets.

## <span style="color:#1f4e79;">Related Problems</span>

* **Prim’s Algorithm:** A vertex-centric MST approach.
* **Disjoint Set Union (DSU):** The underlying data structure for Kruskal's.
* **Cycle Detection:** Using Union-Find to detect cycles in an undirected graph.
* **Borůvka’s Algorithm:** Parallel MST construction.
* **Maximum Spanning Tree:** Finding a spanning tree with maximum total weight.

## <span style="color:#1f4e79;">Use Cases</span>

- **Network Design:** Minimizing costs when laying down cables for telecommunications or power grids.
- **Transportation:** Planning a road or railway network that connects all cities with the least amount of asphalt or track.
- **Image Processing:** Image segmentation where pixels are nodes and edge weights represent color differences.
- **Clustering:** Constructing an MST to identify clusters in data points for machine learning.
- **Approximate TSP:** Kruskal's MST can be used as a basis for approximating the Traveling Salesperson Problem.
