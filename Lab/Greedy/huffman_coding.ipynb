# <span style="color:#1f4e79;">Huffman Tree / Huffman Coding — Greedy Algorithm</span>

**Huffman Coding** is a core greedy algorithm used for **lossless data compression**. It assigns variable-length binary codes to characters based on their frequencies. The primary goal is to minimize the total number of bits required to store or transmit a message by giving shorter codes to more frequent characters and longer codes to rare ones.

Where:  
- `n = number of distinct characters`  
- `char[] = set of characters`  
- `freq[] = frequency of each character`  
- **Prefix Property:** No code is a prefix of another (ensuring unambiguous decoding).

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`Characters:  a,  b,  c,  d,  e,  f`  
`Frequencies: 5,  9, 12, 13, 16, 45`  

Output:  
`f: 0, c: 100, d: 101, a: 1100, b: 1101, e: 111`  

Explanation:  
`f` appears 45 times, so it gets the shortest code (`0`). Characters like `a` and `b` appear rarely, so they are deeper in the tree with longer codes. 



---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`Characters: A, B, C`  
`Frequencies: 10, 15, 30`  

Output:  
`C: 0, A: 10, B: 11`  

Explanation:  
First, A and B (lowest frequencies) are merged into a parent node with frequency 25. Then, this parent is merged with C (frequency 30). Total weighted path length is minimized.
