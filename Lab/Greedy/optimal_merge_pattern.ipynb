# <span style="color:#1f4e79;">Optimal Merge Pattern — Greedy Algorithm</span>

The **Optimal Merge Pattern** is a classic greedy algorithm problem that focuses on finding the most efficient way to merge multiple files into a single file. Each file has a specific size, and merging two files incurs a cost equal to the sum of their sizes. The goal is to determine the minimum total cost required to merge all files into one by always choosing the smallest available files to merge first.

Where:  
- `n = number of files`  
- `files[] = array of file sizes`  
- **Cost of merging** = $size(\text{file}_1) + size(\text{file}_2)$

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`files = [20, 30, 10]`  

Output:  
`90`  

Explanation:  
1. **Initial files:** {10, 20, 30}
2. **Merge 10 and 20:** Cost = 30. Files become {30, 30}.
3. **Merge 30 and 30:** Cost = 60. Final file size = 60.
4. **Total Cost:** $30 + 60 = 90$.



---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`files = [5, 10, 20, 30]`  

Output:  
`115`  

Explanation:  
1. Merge 5 and 10: Cost 15. Files: {15, 20, 30}
2. Merge 15 and 20: Cost 35. Files: {35, 30}
3. Merge 35 and 30: Cost 65.
4. Total Cost: $15 + 35 + 65 = 115$.
