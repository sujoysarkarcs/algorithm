# <span style="color:#1f4e79;">Fractional Knapsack Problem — Greedy Algorithm</span>

The **Fractional Knapsack Problem** is a classic greedy algorithm problem where items can be divided into fractions. The objective is to maximize total value placed in a knapsack with a fixed weight capacity **W**. Unlike the 0/1 Knapsack, you are allowed to take fractions of items, making this problem solvable optimally using a greedy strategy based on **value-to-weight ratio**.

Where:  
- `n = number of items`  
- `W = maximum capacity of the knapsack`  
- `wt[] = array of item weights`  
- `val[] = array of item values`  
- **Note:** Items can be taken partially or fully.

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`weights = [10, 20, 30]`  
`values = [60, 100, 120]`  
`W = 50`  

Output:  
`240.0`  

Explanation:  
1. Calculate value/weight ratio: `[6, 5, 4]`.
2. Take item 1 completely (10kg, 60 value). Remaining `W = 40`.
3. Take item 2 completely (20kg, 100 value). Remaining `W = 20`.
4. Take 2/3 of item 3 (20kg, 80 value). Remaining `W = 0`.
5. Total Value: $60 + 100 + 80 = 240$.

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`weights = [5, 10, 15]`  
`values = [10, 30, 20]`  
`W = 20`  

Output:  
`50.0`  

Explanation:  
Ratio: `[2, 3, 1.33]`. Take item 2 (10kg, 30 val), item 1 (5kg, 10 val), and 1/3 of item 3 (5kg, 10 val).
