# <span style="color:#1f4e79;">Prim’s Minimum Spanning Tree (MST) — Greedy Algorithm</span>

**Prim’s Algorithm** is a classic greedy graph algorithm used to find a **Minimum Spanning Tree (MST)** of a connected, weighted, and undirected graph. An MST is a subset of edges that connects all vertices without any cycles and with the minimum possible total edge weight.

Where:  
- `V = number of vertices`  
- `E = number of edges`  
- `G(V, E) = weighted, undirected graph`  
- **Result:** The MST contains exactly **V − 1** edges.

---

## <span style="color:#1f4e79;">Example 1:</span>

Input (Graph):  
`Vertices: 5`  
`Edges: (0-1, 2), (0-3, 6), (1-2, 3), (1-3, 8), (1-4, 5), (2-4, 7), (3-4, 9)`  

Output:  
`Minimum cost of MST = 16`  

Explanation:  
The edges included in the MST are (0-1), (1-2), (1-4), and (0-3).  
Total weight: $2 + 3 + 5 + 6 = 16$.



---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`Vertices: 3`  
`Edges: (0-1, 1), (1-2, 2), (0-2, 3)`  

Output:  
`Minimum cost of MST = 3`  

Explanation:  
To connect all 3 vertices with minimum cost, we pick edges (0-1) and (1-2).  
Total weight: $1 + 2 = 3$.

- **Network Design:** Laying down fiber optic cables or water pipelines with the lowest possible material cost.
