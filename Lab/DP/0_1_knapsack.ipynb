# <span style="color:#1f4e79;">0/1 Knapsack Problem</span>

Given **n** items, each with a specific **weight** and **value**, and a knapsack of capacity **W**. Determine the maximum total value you can carry without exceeding the weight limit. In the **0/1** version, you cannot break items; you either take an item (1) or leave it (0).


Where:  
- `n = number of items`  
- `W = maximum weight capacity`  
- `wt[] = array of item weights`  
- `val[] = array of item values`

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`n = 4, W = 7`  
`weights = [1, 4, 3, 5]`  
`values = [1500, 3000, 2000, 4000]`

Output:  
`Max profit : 5500`  
`Items taken : Item - 3, Item - 2` (Weights 3 + 4 = 7)

Explanation:  
By picking Item 2 (weight 4, value 3000) and Item 3 (weight 3, value 2000), we get a total value of 5000. However, picking Item 1 and Item 2 gives 4500. The optimal combination for capacity 7 in this case is Item 2 and Item 3.

---

## <span style="color:#1f4e79;">Example 2 (House Robber Analogy):</span>

Input:  
`nums = [2, 7, 9, 3, 1]`  

Output:  
`12`

Explanation:  
Similar to Knapsack, you decide whether to "take" or "not take" a house. Rob house 1 (2), house 3 (9), and house 5 (1), total = $2 + 9 + 1 = 12$.

## <span style="color:#1f4e79;">Core Logic (C snippet)</span>

The transition formula used in the DP table is:
$$dp[i][j] = \max(dp[i-1][j], val[i-1] + dp[i-1][j-wt[i-1]])$$

- **Take:** `val[i-1] + dp[i-1][j-wt[i-1]]` (Current value + value of remaining capacity)
- **Not Take:** `dp[i-1][j]` (Value from previous items for same capacity)
