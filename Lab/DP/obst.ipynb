# <span style="color:#1f4e79;">Optimal Binary Search Tree (OBST) — Dynamic Programming</span>

The **Optimal Binary Search Tree (OBST)** problem focuses on constructing a Binary Search Tree (BST) from a given set of sorted keys such that the **expected search cost** is minimized. Each key has a search frequency (probability) representing how often it is accessed. The goal is to arrange the keys so that frequently searched keys appear closer to the root, reducing the overall search cost.

Where:  
- `n = number of keys`  
- `keys[] = sorted array of keys`  
- `freq[] = array of search frequencies`  
- **Search cost** = $\sum (\text{frequency}_i \times \text{level}_i)$ (where level of root = 1)

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`keys = [10, 12, 20]`  
`freq = [34, 8, 50]`  

Output:  
`142`  

Explanation:  
There are several possible BSTs. Placing key **20** as the root, **10** as its left child, and **12** as the right child of 10 minimizes the cost. High-frequency keys (like 20) are kept at higher levels (closer to the root).

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`keys = [10, 20, 30, 40]`  
`freq = [4, 2, 6, 3]`  

Output:  
`26`
