# <span style="color:#1f4e79;">Staircase Problem — Dynamic Programming</span>

The **Staircase Problem** is a classic counting problem where you are given a staircase with **n steps** and are allowed to climb a certain number of steps at a time (commonly 1 or 2). The objective is to determine the total number of distinct ways to reach the top. This problem is a foundational example of **Dynamic Programming** and shares a direct mathematical relationship with the Fibonacci sequence.

Where:  
- `n = total number of steps`  
- `Allowed moves = 1 step or 2 steps at a time`  
- `Output = total number of distinct ways`

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`n = 2`  

Output:  
`2`  

Explanation:  
There are two ways to reach the top:
1. One step + One step (1 + 1)
2. Two steps at once (2)

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`n = 3`  

Output:  
`3`  

Explanation:  
There are three ways to reach the top:
1. (1 + 1 + 1)
2. (1 + 2)
3. (2 + 1)
