# <span style="color:#1f4e79;">Matrix Chain Multiplication (MCM) — Dynamic Programming</span>

The **Matrix Chain Multiplication** problem deals with finding the most efficient way to multiply a sequence of matrices. The efficiency is measured by the **minimum number of scalar multiplications** required. Since matrix multiplication is associative, the order of multiplication (parenthesization) can be changed, and different orders result in vastly different computational costs.

Where:  
- `n = number of matrices`  
- `arr[] = array of dimensions`  
- `Matrix i` has dimensions `arr[i-1] × arr[i]`

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`arr = [40, 20, 30, 10, 30]`  

Output:  
`26000`  

Explanation:  
There are 4 matrices: $A_1(40\times20), A_2(20\times30), A_3(30\times10), A_4(10\times30)$.  
The optimal parenthesization $(A_1(A_2 A_3))A_4$ results in the minimum cost of 26,000 scalar multiplications.

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`arr = [10, 20, 30]`  

Output:  
`6000`  

Explanation:  
There are 2 matrices: $A_1(10\times20)$ and $A_2(20\times30)$.  
Cost = $10 \times 20 \times 30 = 6000$.
