# <span style="color:#1f4e79;">Coin Change Problem</span>

Given an array of **coins** representing different denominations and a total **amount** of money, determine the number of ways to make the change (or the minimum number of coins needed, depending on the variation). This problem demonstrates the use of **Dynamic Programming** to find an optimal solution by building upon solutions to smaller amounts.

Where:  
- `coins[] = array of available denominations`  
- `amount = the target total sum`  
- Input = array of integers and a target integer  
- Output = total number of ways (or minimum coins)

---

### Example 1:
Input:  
amount = 5, coins = [1, 2, 5]  
Output:  
4  
Explanation: There are four ways to make the change:
1. {1, 1, 1, 1, 1}
2. {1, 1, 1, 2}
3. {1, 2, 2}
4. {5}

### Example 2:
Input:  
amount = 3, coins = [2]  
Output:  
0  
Explanation: It is impossible to make the sum of 3 using only coins of denomination 2.
