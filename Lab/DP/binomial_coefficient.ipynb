# <span style="color:#1f4e79;">Binomial Coefficient — Dynamic Programming</span>

The **Binomial Coefficient** represents the number of ways to choose **k** elements from a set of **n** distinct elements without considering the order of selection. It is commonly denoted as $C(n, k)$ or $\binom{n}{k}$. This problem is a fundamental building block in combinatorics and is often solved using **Dynamic Programming** to avoid the overhead of large factorial calculations.

Where:  
- `n = total number of elements`  
- `k = number of elements to choose`  
- `Output = total number of combinations`

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`n = 5, k = 2`  

Output:  
`10`  

Explanation:  
Choosing 2 elements from {A, B, C, D, E} results in 10 pairs: {A,B}, {A,C}, {A,D}, {A,E}, {B,C}, {B,D}, {B,E}, {C,D}, {C,E}, and {D,E}.

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`n = 10, k = 0`  

Output:  
`1`  

Explanation:  
There is exactly one way to choose zero elements from any set (the empty set).
