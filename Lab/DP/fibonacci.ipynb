# <span style="color:#1f4e79;">Fibonacci Numbers</span>

The **Fibonacci sequence** is a mathematical series where each number is the sum of the two preceding ones. It typically starts with **0** and **1**. The sequence is defined by the recurrence relation:  
$F(n) = F(n-1) + F(n-2)$  
with base cases:  
$F(0) = 0$ and $F(1) = 1$

This sequence is a primary example used to demonstrate the power of **Dynamic Programming**, as it transforms an exponential $O(2^n)$ recursive process into a highly efficient $O(n)$ linear process.

---

### Example 1:
Input:  
n = 5  
Output:  
5  
Explanation: The sequence is 0, 1, 1, 2, 3, 5. The value at index 5 is 5.

### Example 2:
Input:  
n = 8  
Output:  
21  
Explanation: The sequence continues: ..., 5, 8, 13, 21. The value at index 8 is 21.

### Use Cases:
- **Algorithm Optimization:** Demonstrating Memoization vs. Tabulation in Dynamic Programming.
- **Natural Patterns:** Modeling growth patterns in plants, such as sunflower seeds and pinecone spirals.
- **Financial Ratios:** Used in "Fibonacci Retracement" to identify potential reversal levels in stock charts.
- **System Design:** Implementing Fibonacci Heaps to optimize priority queue operations in graph algorithms like Dijkstra's.
- **Architecture:** Utilizing the "Golden Ratio" (derived from Fibonacci) for aesthetically pleasing proportions.
