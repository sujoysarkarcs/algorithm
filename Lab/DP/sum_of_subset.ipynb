# <span style="color:#1f4e79;">Sum of Subsets Problem — Dynamic Programming</span>

The **Sum of Subsets Problem** is a classic decision problem where you are given a set of **n positive integers** and a target value **S**. The task is to determine whether there exists at least one subset of the given set whose sum is exactly equal to **S**.  
Each element in the set can either be included once or excluded; elements cannot be reused.

Where:  
- `n = number of elements`  
- `arr[] = array of positive integers`  
- `S = target sum`

---

## <span style="color:#1f4e79;">Example 1:</span>

Input:  
`arr = [3, 34, 4, 12, 5, 2], S = 9`  

Output:  
`True`  

Explanation:  
The subset `{4, 5}` results in a sum of $4 + 5 = 9$.

---

## <span style="color:#1f4e79;">Example 2:</span>

Input:  
`arr = [1, 2, 7, 8], S = 5`  

Output:  
`False`  

Explanation:  
No combination of elements from the set can sum up to exactly 5.

## <span style="color:#1f4e79;">Core Logic (Transition)</span>

The problem is solved using a 2D boolean DP table where `dp[i][j]` represents if a sum `j` is possible using the first `i` elements:

$$dp[i][j] = dp[i-1][j] \text{ OR } dp[i-1][j - arr[i-1]]$$

- **Exclude:** `dp[i-1][j]` (The sum was already possible without the current element)
- **Include:** `dp[i-1][j - arr[i-1]]` (The remaining sum was possible with previous elements)
