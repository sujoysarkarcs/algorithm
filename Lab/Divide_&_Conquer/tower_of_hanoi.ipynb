## Tower of Hanoi — Divide and Conquer

Given **n disks** of different sizes placed on a **source rod**, move all the disks to a **destination rod** using an **auxiliary rod**, following the rules of the Tower of Hanoi problem.

The Tower of Hanoi algorithm works by **recursively dividing** the problem into smaller subproblems:
first moving **n−1 disks** to the auxiliary rod, then moving the **largest disk** to the destination rod, and finally moving the **n−1 disks** from the auxiliary rod to the destination rod.

---

### Where:

* **n** = number of disks
* **Input** = n disks on the source rod
* **Output** = all disks moved to the destination rod in correct order

---

### Example 1:

**Input:**
n = 2
Source = A, Auxiliary = B, Destination = C

**Output (moves):**

1. Move disk 1 from A to B
2. Move disk 2 from A to C
3. Move disk 1 from B to C

**Explanation:**
The problem is divided into moving 1 disk at a time, and the moves are combined to transfer all disks to the destination rod.

---

### Example 2:

**Input:**
n = 3
Source = A, Auxiliary = B, Destination = C

**Output (moves):**

1. Move disk 1 from A to C
2. Move disk 2 from A to B
3. Move disk 1 from C to B
4. Move disk 3 from A to C
5. Move disk 1 from B to A
6. Move disk 2 from B to C
7. Move disk 1 from A to C

**Explanation:**
The problem of moving 3 disks is divided into two subproblems of size 2, solved recursively, and combined by moving the largest disk.
