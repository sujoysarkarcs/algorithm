# Fundamentals of Recurrence Relations

### 1. What is Recurrence?
In computer science, **recurrence** occurs when a problem is solved by breaking it down into smaller, self-similar versions of the same problem.

If a function calls itself to solve a smaller instance of the input, the time complexity cannot be written as a simple equation (like $T(n) = n^2$). Instead, it is expressed as a **Recurrence Relation**: an equation that defines a sequence recursively.

#### Definition
A **Recurrence Relation** is an equation that expresses the time complexity $T(n)$ for an input of size $n$ in terms of $T$ for smaller input sizes (like $T(n-1)$ or $T(n/2)$).

#### Simple Example
Imagine you are standing in a line of people. You want to know which position you are in.
* **Recursive Step:** You ask the person in front of you, "What number are you?" and add 1 to their answer.
* **Base Case:** The person at the front knows they are #1.

Mathematically:
$$
T(n) = T(n-1) + 1
$$

---

### 2. Key Terms in Recurrence
To read a recurrence relation, you must understand its components.

$$
T(n) = aT\left(\frac{n}{b}\right) + f(n)
$$

1.  **$T(n)$:** The total time required to solve the problem of size $n$.
2.  **Recursive Term ($aT(\frac{n}{b})$):** The time taken by the sub-problems.
    * $a$: Number of recursive calls (branches).
    * $\frac{n}{b}$: The new size of the data in each call.
3.  **Non-Recursive Term ($f(n)$):** The extra work done outside the recursion (dividing the problem or merging results).
4.  **Base Case:** The condition where recursion stops (usually $T(1) = 1$ or $T(0) = 0$).

---

### 3. How to Form a Recurrence Relation from Code
To turn code into math, we analyze the cost of each line.

#### Case A: Linear Recursion (Reducing size by subtraction)
**Code:**
```c
void printCountdown(int n) {
    if (n == 0) return;         // Constant time: c
    printf("%d", n);            // Constant time: c
    printCountdown(n - 1);      // Recursive call: T(n-1)
}

**Analysis:**
1.  The base case takes constant time: $O(1)$.
2.  The print operation takes constant time: $O(1)$ (let's call it $1$ unit).
3.  The function calls itself with size $n-1$: this costs $T(n-1)$.

**Relation:**
$$
T(n) = T(n-1) + 1
$$

#### Case B: Divide and Conquer (Reducing size by division)
**Code:**
```c
void unknownAlgo(int n) {
    if (n == 1) return;
    
    // Loop runs n times
    for (int i = 0; i < n; i++) { 
        printf("Processing"); 
    }
    
    // Two recursive calls with half size
    unknownAlgo(n/2);
    unknownAlgo(n/2);
}

## Analysis
The loop runs $n$ times: Cost is $n$.  
The function calls itself twice, each with size $n/2$: Cost is $2T(n/2)$.

**Relation:**
$$T(n) = 2T(n/2) + n$$

---

## 4. Recurrence Relations of Famous Programs
Here are the standard relations for the algorithms you listed. It is crucial to understand why they look this way.

### 1. Factorial
Calculates $n! = n \times (n-1)!$.

**Logic:** One multiplication ($O(1)$) combined with one recursive call for $n-1$.

**Relation:**
$$T(n) = T(n-1) + 1$$

---

### 2. Binary Search
Finds an element in a sorted array by checking the middle.

**Logic:** Compare middle ($O(1)$). If not found, discard half and search the remaining half.

**Relation:**
$$T(n) = T(n/2) + 1$$

---

### 3. Tower of Hanoi
Moves $n$ disks from rod A to rod C.

**Logic:**
- Move $n-1$ disks from A → B ($T(n-1)$)  
- Move largest disk A → C ($1$)  
- Move $n-1$ disks B → C ($T(n-1)$)

**Relation:**
$$T(n) = 2T(n-1) + 1$$

---

### 4. Merge Sort
Sorts an array by splitting it.

**Logic:**
- Sort left half → $T(n/2)$  
- Sort right half → $T(n/2)$  
- Merge → $n$

**Relation:**
$$T(n) = 2T(n/2) + n$$

---

### 5. Quick Sort
Sorts using a pivot and partitioning.

**Best Case (Balanced):**
$$T(n) = 2T(n/2) + n$$

**Worst Case (Skewed):**
$$T(n) = T(n-1) + n$$

---

### 6. Fibonacci (Naive Recursive)
Calculates $F(n) = F(n-1) + F(n-2)$.

**Logic:** Two recursive calls + constant work.

**Relation:**
$$T(n) = T(n-1) + T(n-2) + 1$$

---

## Methods to Solve Recurrence Relations

Recurrence relations are used to express the running time of recursive algorithms. There are several standard techniques to solve them:

### 1. Iterative / Substitution Method
- Expand the recurrence step-by-step.
- Look for a pattern after repeated substitutions.
- Generalize using a variable (usually \(k\)).
- Apply the stopping condition (base case) and simplify.
- Useful for **simple recurrences** like \(T(n)=T(n-1)+1\) or \(T(n)=2T(n/2)+n\).

---

### 2. Recursion Tree Method
- Visualize the recurrence as a **tree of recursive calls**.
- Compute work done at each level.
- Sum work across all levels.
- Helps in understanding **total work distribution**.
- Commonly used for divide-and-conquer recurrences like Merge Sort and Fibonacci.

---

### 3. Master Theorem
Used for recurrences of the form:
\[
T(n)=aT\left(\frac{n}{b}\right)+f(n)
\]
where:
- \(a\) = number of subproblems  
- \(b\) = size reduction factor  
- \(f(n)\) = work done outside recursion  

Gives direct asymptotic complexity without full expansion.  
Best suited for algorithms like Merge Sort, Binary Search Tree operations, Strassen’s algorithm, etc.

---

### 4. Substitution / Mathematical Induction
- Guess the solution form.
- Prove it using **induction**.
- Often used to **verify** results from other methods.
- Helpful when Master Theorem cannot be directly applied.

---

### 5. Characteristic Equation Method
- Used for **linear homogeneous recurrences** such as:
\[
T(n)=aT(n-1)+bT(n-2)
\]
- Solve using algebraic roots.
- Commonly applied to Fibonacci-type recurrences.

---

### Summary
| Method | Best For | Examples |
|-------|---------|---------|
| Iteration | Simple expansions | \(T(n)=T(n-1)+1\) |
| Recursion Tree | Visualizing levels of work | Merge Sort, Fibonacci |
| Master Theorem | Divide-and-conquer | \(T(n)=2T(n/2)+n\) |
| Induction Proof | Validating solutions | Any recurrence |
| Characteristic Equation | Linear recurrences | Fibonacci, DP relations |


## The Iterative Method

### 1. The 4-Step Process
To solve any recurrence using iteration, follow these four logical steps:

1. **Expand** — Substitute the recursive term ($T(n-1)$ or $T(n/2)$) into the original equation 2–3 times.  
2. **Generalize (The $k$-th Step)** — Observe the pattern after $k$ expansions and write a general form.  
3. **Find the Stopping Condition** — Determine the value of $k$ where the input reaches the base case.  
4. **Solve** — Substitute $k$ back into the general form to obtain the time complexity.

---

## 2. Simple Example: Factorial

**Recurrence:** $T(n) = T(n-1) + 1$  
**Base Case:** $T(0)=1$

### Step 1: Expand
Original:
$$T(n) = T(n-1) + 1$$

Substitute $T(n-1) = T(n-2) + 1$:
$$T(n) = [T(n-2) + 1] + 1 \implies T(n-2) + 2$$

Substitute $T(n-2) = T(n-3) + 1$:
$$T(n) = [T(n-3) + 1] + 2 \implies T(n-3) + 3$$

### Step 2: Generalize
$$T(n) = T(n-k) + k$$

### Step 3: Stopping Condition
$$n-k = 0 \implies k = n$$

### Step 4: Solve
$$T(n) = T(0) + n = 1 + n \implies \mathbf{O(n)}$$

---

## 3. Detailed Solutions for Standard Algorithms

### A. Binary Search (Divide by 2)

**Recurrence:**  
$$T(n) = T\left(\frac{n}{2}\right) + c$$  

**Base Case:** $T(1)=c$

#### Expand
$$T(n) = T\left(\frac{n}{2}\right) + c$$

$$T(n) = \left[ T\left(\frac{n}{4}\right) + c \right] + c
= T\left(\frac{n}{2^2}\right) + 2c$$

$$T(n) = \left[ T\left(\frac{n}{8}\right) + c \right] + 2c
= T\left(\frac{n}{2^3}\right) + 3c$$

#### Generalize
$$T(n) = T\left(\frac{n}{2^k}\right) + k \cdot c$$

#### Stopping Condition
$$\frac{n}{2^k} = 1 \implies 2^k = n \implies k = \log_2 n$$

#### Solve
$$T(n) = T(1) + (\log_2 n)\cdot c \implies \mathbf{O(\log n)}$$

---

### B. Merge Sort (Divide by 2 + Linear Work)

**Recurrence:**  
$$T(n) = 2T\left(\frac{n}{2}\right) + n$$  
**Base Case:** $T(1)=1$

#### Expand
$$T(n) = 2T\left(\frac{n}{2}\right) + n$$

Substitute $T(n/2) = 2T(n/4) + n/2$:
$$T(n) = 2\left[ 2T\left(\frac{n}{4}\right) + \frac{n}{2} \right] + n$$
$$T(n) = 4T\left(\frac{n}{4}\right) + n + n = 4T\left(\frac{n}{4}\right) + 2n$$

Substitute $T(n/4) = 2T(n/8) + n/4$:
$$T(n) = 4\left[ 2T\left(\frac{n}{8}\right) + \frac{n}{4} \right] + 2n$$
$$T(n) = 8T\left(\frac{n}{8}\right) + n + 2n = 8T\left(\frac{n}{8}\right) + 3n$$

#### Generalize
$$T(n) = 2^k T\left(\frac{n}{2^k}\right) + k \cdot n$$

#### Stopping Condition
$$\frac{n}{2^k} = 1 \implies k = \log_2 n$$

#### Solve
$$T(n) = n + n\log n \implies \mathbf{O(n\log n)}$$

---

### C. Tower of Hanoi (Two branches, subtract 1)

**Recurrence:**  
$$T(n) = 2T(n-1) + 1$$  
**Base Case:** $T(0)=0$ or $T(1)=1$

#### Expand
$$T(n) = 2T(n-1) + 1$$
$$T(n) = 2[2T(n-2) + 1] + 1 = 4T(n-2) + 2 + 1$$
$$T(n) = 4[2T(n-3) + 1] + 2 + 1 = 8T(n-3) + 4 + 2 + 1$$

#### Generalize
$$T(n) = 2^k T(n-k) + (2^{k-1} + \dots + 2 + 1)$$

Stopping at $k=n$:
$$T(n) = 2^n - 1 \implies \mathbf{O(2^n)}$$

---

### D. Quick Sort (Worst Case)

**Recurrence:**  
$$T(n) = T(n-1) + n$$

#### Expand
$$T(n) = T(n-1) + n$$
$$T(n) = [T(n-2) + (n-1)] + n$$
$$T(n) = [T(n-3) + (n-2)] + (n-1) + n$$

#### Generalize
$$T(n) = T(n-k) + \sum_{i=0}^{k-1} (n-i)$$

Stopping at $k=n-1$:
$$T(n) = \frac{n^2 + n}{2} \implies \mathbf{O(n^2)}$$

---

## Note on Fibonacci
The recurrence  
$$T(n) = T(n-1) + T(n-2) + 1$$  
reduces at different rates, so iteration is unsuitable.  
We instead use the **recursion tree** or **characteristic equation**, giving:
$$O(2^n)$$


## Strassen’s Matrix Multiplication

### 1. The Problem: Standard Matrix Multiplication
Given two $n \times n$ matrices $A$ and $B$, we want to compute their product $C = A \times B$.

In the standard row–column method:

- Each element of $C$ requires $n$ multiplications and $n-1$ additions  
- There are $n^2$ elements  

So the total time complexity is:

$$O(n^3)$$

---

### 2. Naive Divide and Conquer

We divide each matrix into four $\frac{n}{2} \times \frac{n}{2}$ submatrices:

$$
A =
\begin{bmatrix}
A_{11} & A_{12} \\
A_{21} & A_{22}
\end{bmatrix},
\quad
B =
\begin{bmatrix}
B_{11} & B_{12} \\
B_{21} & B_{22}
\end{bmatrix}
$$

The result matrix $C$ is computed as:

$$C_{11} = A_{11}B_{11} + A_{12}B_{21}$$
$$C_{12} = A_{11}B_{12} + A_{12}B_{22}$$
$$C_{21} = A_{21}B_{11} + A_{22}B_{21}$$
$$C_{22} = A_{21}B_{12} + A_{22}B_{22}$$

#### Analysis of Naive D&C
- $8$ recursive multiplications of size $n/2$  
- $4$ matrix additions of size $(n/2) \times (n/2)$, which take $O(n^2)$ time  

The recurrence is:

$$T(n) = 8T\left(\frac{n}{2}\right) + O(n^2)$$

Using Master Theorem, $\log_2 8 = 3$, therefore:

$$T(n) = O(n^3)$$

(No improvement over the standard method.)

---

### 3. Strassen’s Algorithm (Improvement)

In 1969, Volker Strassen reduced the number of multiplications from $8$ to $7$ by performing extra additions and subtractions (which are cheaper).

#### The 7 Intermediate Products

$$P_1 = A_{11}(B_{12} - B_{22})$$
$$P_2 = (A_{11} + A_{12})B_{22}$$
$$P_3 = (A_{21} + A_{22})B_{11}$$
$$P_4 = A_{22}(B_{21} - B_{11})$$
$$P_5 = (A_{11} + A_{22})(B_{11} + B_{22})$$
$$P_6 = (A_{12} - A_{22})(B_{21} + B_{22})$$
$$P_7 = (A_{11} - A_{21})(B_{11} + B_{12})$$

#### Reconstructing the Result

$$C_{11} = P_5 + P_4 - P_2 + P_6$$
$$C_{12} = P_1 + P_2$$
$$C_{21} = P_3 + P_4$$
$$C_{22} = P_5 + P_1 - P_3 - P_7$$

---

### 4. Solving the Recurrence

Recursive work: $7T(n/2)$  
Non-recursive work: $\Theta(n^2)$

$$T(n) = 7T\left(\frac{n}{2}\right) + c n^2$$

Using Master Theorem:

- $a = 7$
- $b = 2$
- $f(n) = n^2$
- $\log_2 7 \approx 2.81$

Since $n^{2.81} > n^2$, the recursive term dominates.

Final complexity:

$$T(n) = \Theta(n^{\log_2 7}) \approx \Theta(n^{2.81})$$

---

### 5. Summary Comparison

| Algorithm | Multiplications | Recurrence | Complexity |
|----------|----------------|-----------|-----------|
| Naive Divide & Conquer | 8 | $T(n) = 8T(n/2) + n^2$ | $O(n^3)$ |
| Strassen | 7 | $T(n) = 7T(n/2) + n^2$ | $O(n^{2.81})$ |
