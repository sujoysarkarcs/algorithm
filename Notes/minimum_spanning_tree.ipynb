# Minimum Spanning Tree (MST)

---

## 1. What is a Minimum Spanning Tree?

A **Minimum Spanning Tree (MST)** is a subset of edges of a connected, weighted graph such that:

- It connects all vertices
- It has **no cycles**
- The **total edge weight is minimum**

---

## 2. Key Properties

- For a graph with **V vertices**, MST has **V-1 edges**
- No cycles are allowed
- There can be multiple MSTs if weights are equal

---

## 3. Algorithms to Find MST

Two most important algorithms:

1. **Prim’s Algorithm** (Greedy, vertex-based)
2. **Kruskal’s Algorithm** (Greedy, edge-based)

---

# 4. Prim’s Algorithm

## 4.1 Idea

- Start from any vertex
- Grow the tree by adding the **minimum weight edge**
- Always pick edge connecting:
  - a visited vertex
  - an unvisited vertex

---

## 4.2 Working Steps

1. Initialize all keys as ∞
2. Pick starting vertex → key = 0
3. Select minimum key vertex not in MST
4. Update adjacent vertices
5. Repeat until all vertices included

---

## 4.3 Code (Prim’s Algorithm)

```c id="prim_code_block"
#include<stdio.h>
#include <limits.h>
#include <stdbool.h>
#define V 5

void printMST(int parent[], int graph[V][V]) {
    printf("Edge \tWeight\n");
    int totalWeight = 0;
    for (int i = 1; i < V; i++) {
        printf("%d - %d \t%d \n", parent[i], i, graph[i][parent[i]]);
        totalWeight += graph[i][parent[i]];
    }
    printf("Total Cost: %d\n", totalWeight);
}

int minKey(int key[], bool mstSet[]){
  int min=INT_MAX, min_index;
  for(int v=0; v<V; v++){
    if(mstSet[v] == 0 && key[v] < min){
      min = key[v];
      min_index = v;
    }
  }
  return min_index;
}

void primMST(int graph[V][V]){
  int parent[V];
  int key[V];
  bool mstSet[V];

  for(int v=0; v<V; v++){
    mstSet[v] = 0;
    key[v] = INT_MAX;
  }

  parent[0] = -1;
  key[0] = 0;

  for(int count=0; count < V-1; count++){
    int u = minKey(key, mstSet);
    mstSet[u] = 1;

    for(int v=0; v < V; v++){
      if(graph[u][v] > 0 && mstSet[v] == 0 && graph[u][v] < key[v]){
        parent[v] = u;
        key[v] = graph[u][v];
      }
    }
  }

  printMST(parent, graph);
}

int main() {
    int graph[V][V] = {
        {0, 2, 0, 6, 0},
        {2, 0, 3, 8, 5},
        {0, 3, 0, 0, 7},
        {6, 8, 0, 0, 9},
        {0, 5, 7, 9, 0}
    };

    primMST(graph);

    return 0;
}
````

---

## 4.4 Explanation

* `key[]` → stores minimum weight to connect vertex
* `mstSet[]` → tracks included vertices
* `parent[]` → stores MST structure

### Flow:

* Pick smallest key
* Add to MST
* Update neighbors

---

## 4.5 Time Complexity

* Using adjacency matrix: **O(V²)**

---

# 5. Kruskal’s Algorithm

## 5.1 Idea

* Sort all edges by weight
* Pick smallest edge
* Add if it **does not form a cycle**

---

## 5.2 Key Concept: Union-Find

Used to detect cycles:

* `find()` → finds root
* `union()` → merges sets

---

## 5.3 Code (Kruskal’s Algorithm)

```c id="kruskal_code_block"
#include <stdio.h>
#include <limits.h>

#define MAX 9

int find(int parent[], int v)
{
    while (parent[v] != v)
        v = parent[v];
    return v;
}

int unite(int parent[], int r1, int r2)
{
    if (r1 != r2) {
        parent[r2] = r1;
        return 1;
    }
    return 0;
}

void mst(int n, int cost[3][3])
{
    int parent[MAX];

    for (int i = 0; i < n; i++)
        parent[i] = i;

    int cnt = 1;
    int sum = 0;

    printf("MST Edges:\n");

    while (cnt < n) {

        int min = INT_MAX;
        int a = -1, b = -1;

        for (int i = 0; i < n; i++) {
            for (int j = 0; j < n; j++) {
                if (cost[i][j] < min) {
                    min = cost[i][j];
                    a = i;
                    b = j;
                }
            }
        }

        int r1 = find(parent, a);
        int r2 = find(parent, b);

        if (unite(parent, r1, r2)) {
            printf("%d - %d\n", a, b);
            sum += min;
        }

        cost[a][b] = cost[b][a] = INT_MAX;
        cnt++;

        for(int i=0; i<MAX; i++){
            printf("%d-%d ", i, parent[i]);
        }
        printf("\n");
    }

    printf("\nTotal Cost = %d\n", sum);
}

int main()
{
    int n = 3;

    int cost[3][3] = {
        {0, 10, 20},
        {12, 0, 15},
        {16, 18, 0}
    };

    for (int i = 0; i < n; i++)
        for (int j = 0; j < n; j++)
            if (cost[i][j] == 0)
                cost[i][j] = INT_MAX;

    mst(n, cost);

    return 0;
}
```

---

## 5.4 Explanation

* Select minimum edge globally
* Check if it forms cycle using `find()`
* If not → include in MST

### Important:

* `parent[]` represents disjoint sets
* Prevents cycles

---

## 5.5 Time Complexity

* Sorting edges: **O(E log E)**
* Union-Find: nearly constant

---

# 6. Prim vs Kruskal

| Feature        | Prim’s Algorithm | Kruskal’s Algorithm |
| -------------- | ---------------- | ------------------- |
| Approach       | Vertex-based     | Edge-based          |
| Data Structure | Priority/Matrix  | Union-Find          |
| Best for       | Dense Graphs     | Sparse Graphs       |
| Cycle Check    | Not needed       | Required            |

---

# 7. Summary

* **Prim’s** grows MST from a starting node
* **Kruskal’s** builds MST by selecting smallest edges
* Both are **Greedy Algorithms**

---

## 🔑 One-Line Summary

> MST = Connect all vertices with minimum cost and no cycles
