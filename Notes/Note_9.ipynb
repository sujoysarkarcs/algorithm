# 1️⃣ Fractional Knapsack Problem

<div style="background-color:#E8F6F3; padding:15px; border-radius:10px;">

### 📌 Definition
You have a knapsack with capacity $W$.  
Each item has a **weight** and a **value**.

✅ Unlike 0/1 Knapsack, here you can **break items into fractions**.

### 🧠 Greedy Choice
Always pick the item with the highest value per unit weight (density):

$$
\text{Density} = \frac{\text{Value}}{\text{Weight}}
$$

</div>

---

### 🔢 Numerical Example

- Capacity = 50 kg  
- Item 1: Value = 60, Weight = 10 → Density = $6$  
- Item 2: Value = 100, Weight = 20 → Density = $5$  
- Item 3: Value = 120, Weight = 30 → Density = $4$  

---

### ✅ Solution

**Step 1:** Pick Item 1 (Highest density = 6)  
Remaining capacity = 40  
Total value = 60  

**Step 2:** Pick Item 2 (Density = 5)  
Remaining capacity = 20  
Total value = 160  

**Step 3:** Pick fraction of Item 3  

$$
\text{Fraction taken} = \frac{20}{30}
$$

$$
\text{Value added} = 120 \times \frac{20}{30} = 80
$$

### 🏆 Final Answer

$$
\textbf{Maximum Value} = 240
$$

---
### 📝 Algorithm: Fractional Knapsack

<div style="background-color:#F4F6F7; padding:15px; border-radius:10px; border-left: 5px solid #2C3E50;">

**Input:** A set $I$ of $n$ items, where each item $i$ has value $i.value$ and weight $i.weight$. A maximum knapsack capacity $W$.

**Output:** The maximum total value that can be carried in the knapsack.

```text
Algorithm FRACTIONAL-KNAPSACK(I, W)

    n ← |I|

    FOR i ← 1 TO n DO
        I[i].density ← I[i].value / I[i].weight
    END FOR

    SORT(I) such that
        I[1].density ≥ I[2].density ≥ ... ≥ I[n].density

    total_value ← 0
    remaining_capacity ← W

    FOR i ← 1 TO n DO

        IF I[i].weight ≤ remaining_capacity THEN

            remaining_capacity ← remaining_capacity - I[i].weight
            total_value ← total_value + I[i].value

        ELSE

            fraction ← remaining_capacity / I[i].weight
            total_value ← total_value + (I[i].value × fraction)

            remaining_capacity ← 0
            EXIT FOR

        END IF

    END FOR

    RETURN total_value
```

# 2️⃣ Huffman Coding

---

## 1️⃣ The Problem: Saving Space

<div style="background-color:#E8F6F3; padding:15px; border-radius:10px;">

Computers store text as 0s and 1s (bits). Normally, a computer uses the exact same number of bits for every single letter. For example, it might use 3 bits for every letter.

But think about the English language:

- We use the letter **'e'** very frequently.
- We rarely use the letter **'z'**.

If we use the exact same amount of memory for 'e' as we do for 'z', we are wasting space.

### 🎯 The Goal of Huffman Coding
To save memory by creating a **custom code** for our message.

- Very short codes → Frequently used characters  
- Longer codes → Rarely used characters  

(Think of it like Morse Code.)

We do this **without losing any part of the original message**.

</div>

---

## 2️⃣ The Golden Rule (Prefix Rule)

<div style="background-color:#FEF9E7; padding:15px; border-radius:10px;">

If we start making our own custom codes of different lengths  
(for example, 0 for 'e' and 01 for 's'), the computer will get confused.

If it sees 01:

- Is that an 'e' (0) followed by something else?
- Or is it an 's' (01)?

### 🔒 Prefix Rule
No character's code can be the start (prefix) of another character's code.

This guarantees **unambiguous decoding**.

</div>

---

# 2️⃣ Huffman Coding

<div style="background-color:#FEF9E7; padding:15px; border-radius:10px;">

### 📌 Definition
Huffman Coding is a lossless data compression algorithm. It assigns variable-length binary codes to characters based on their frequency of occurrence. Characters that appear frequently are assigned shorter codes, while characters that appear rarely receive longer codes.

### 🧠 The Greedy Choice
Always select and merge the two nodes with the lowest frequencies to build a binary tree from the bottom up.

</div>

---

## 🔢 Numerical Example

### Given Data
A message contains 100 characters with the following frequencies:

| Character | Frequency |
|-----------|-----------|
| A | 5 |
| B | 9 |
| C | 12 |
| D | 13 |
| E | 16 |
| F | 45 |

---

# ✅ Step-by-Step Solution

## Part 1: Building the Tree

Initialize the pool of available nodes:

$$
[A:5,\ B:9,\ C:12,\ D:13,\ E:16,\ F:45]
$$

---

### 1️⃣ Merge the two smallest (5 and 9)

$$
5 + 9 = 14
$$

Create a new internal node (14).

Pool becomes:

$$
[C:12,\ D:13,\ Node:14,\ E:16,\ F:45]
$$

---

### 2️⃣ Merge the two smallest (12 and 13)

$$
12 + 13 = 25
$$

Create a new internal node (25).

Pool becomes:

$$
[Node:14,\ E:16,\ Node:25,\ F:45]
$$

---

### 3️⃣ Merge the two smallest (14 and 16)

$$
14 + 16 = 30
$$

Create a new internal node (30).

Pool becomes:

$$
[Node:25,\ Node:30,\ F:45]
$$

---

### 4️⃣ Merge the two smallest (25 and 30)

$$
25 + 30 = 55
$$

Create a new internal node (55).

Pool becomes:

$$
[F:45,\ Node:55]
$$

---

### 5️⃣ Merge the final two (45 and 55)

$$
45 + 55 = 100
$$

This is the Root node.  
The tree is complete.

---

## Part 2: Assigning Codes

Starting from the root node:

- Assign **0** to every left edge  
- Assign **1** to every right edge  

Trace the path from the root to each leaf node to determine its code.

---

| Character | Frequency | Path | Huffman Code | Code Length |
|------------|------------|-------|---------------|-------------|
| **F** | 45 | Left | **0** | 1 bit |
| **E** | 16 | Right $\rightarrow$ Right | **11** | 2 bits |
| **C** | 12 | Right $\rightarrow$ Left $\rightarrow$ Left | **100** | 3 bits |
| **D** | 13 | Right $\rightarrow$ Left $\rightarrow$ Right | **101** | 3 bits |
| **A** | 5 | Right $\rightarrow$ Right $\rightarrow$ Left $\rightarrow$ Left | **1100** | 4 bits |
| **B** | 9 | Right $\rightarrow$ Right $\rightarrow$ Left $\rightarrow$ Right | **1101** | 4 bits |

# 📊 Compression Analysis

To calculate the efficiency of the algorithm, compare the total bits required using standard fixed-length codes versus the new variable-length Huffman codes.

---

## 1️⃣ Standard Fixed-Length Encoding

6 unique characters require a minimum of 3 bits per character:

$$
2^3 = 8
$$

Total size:

$$
100 \times 3 = 300 \text{ bits}
$$

---

## 2️⃣ Huffman Encoding

Multiply each character's frequency by its new code length:

- F: $45 \times 1 = 45$ bits  
- E: $16 \times 2 = 32$ bits  
- C: $12 \times 3 = 36$ bits  
- D: $13 \times 3 = 39$ bits  
- A: $5 \times 4 = 20$ bits  
- B: $9 \times 4 = 36$ bits  

Total size:

$$
45 + 32 + 36 + 39 + 20 + 36 = 208 \text{ bits}
$$

---

# 🏆 Result

The Huffman Coding algorithm compressed the data from 300 bits to 208 bits.

---
### 📝 The Algorithm (Formal Notation)

<div style="background-color:#F4F6F7; padding:15px; border-radius:10px; border-left: 5px solid #2C3E50;">

**Input:** A set $C$ of $n$ characters, where each character $c \in C$ has frequency $c.freq$  

**Output:** The root node of the Huffman tree

```text
Algorithm HUFFMAN(C)

    n ← |C|
    Q ← Build-Min-Priority-Queue(C)

    FOR i ← 1 TO n - 1 DO

        Allocate a new node z

        z.left  ← x ← EXTRACT-MIN(Q)
        z.right ← y ← EXTRACT-MIN(Q)

        z.freq ← x.freq + y.freq

        INSERT(Q, z)

    END FOR

    RETURN EXTRACT-MIN(Q)
```

---

**Input:** The root node of the Huffman tree and an empty string `code`

```text
Algorithm GENERATE-CODES(node, code)

    IF node is a leaf THEN
        PRINT node.char, " : ", code
        RETURN
    END IF

    IF node.left ≠ NULL THEN
        GENERATE-CODES(node.left, code + "0")
    END IF

    IF node.right ≠ NULL THEN
        GENERATE-CODES(node.right, code + "1")
    END IF
```

# 📘 Chapter: Optimal Merge Pattern

---

## 1️⃣ The Problem: The Cost of Merging

<div style="background-color:#E8F6F3; padding:15px; border-radius:10px;">

Imagine you have multiple sorted files (like lists of student grades) and you need to combine them into one massive sorted file.

To merge two files, the computer has to look at every single record in both files. If File A has $m$ records and File B has $n$ records, the cost (number of operations) to merge them is $m + n$.

When you have more than two files, the order in which you merge them completely changes the total amount of work the computer has to do. If you merge a massive file too early, its records will get dragged into every subsequent merge, wasting a huge amount of processing power.

</div>

---

## 2️⃣ The Greedy Strategy

<div style="background-color:#FEF9E7; padding:15px; border-radius:10px;">

To minimize the total number of record movements, we use a bottom-up Greedy approach.

**The Rule:** Always pick the **two smallest files** available, merge them, and put the resulting new file back into your pool. Repeat this until only one file remains.

</div>

---

## 3️⃣ Numerical Example & Step-by-Step Solution

**Given Data:**  
We have 4 sorted files with the following sizes (number of records):

- $L_1 = 2$
- $L_2 = 3$
- $L_3 = 4$
- $L_4 = 5$

**Initial Pool:**

$$
\{2, 3, 4, 5\}
$$

---

### Step 1: Merge the two smallest (2 and 3)

- We merge $L_1$ and $L_2$.
- The cost of this merge is:

$$
2 + 3 = \mathbf{5}
$$

- We remove 2 and 3, and add the new file (size 5) to our pool.

**Current Pool:**

$$
\{4, 5, 5\}
$$

(Note: one 5 is $L_4$, the other is our new merged file.)

---

### Step 2: Merge the two smallest (4 and 5)

- We merge $L_3$ (4) and one of the 5s.
- The cost of this merge is:

$$
4 + 5 = \mathbf{9}
$$

- We remove 4 and 5, and add the new file (size 9) to our pool.

**Current Pool:**

$$
\{5, 9\}
$$

---

### Step 3: Merge the final two (5 and 9)

- We merge the remaining file of size 5 and the file of size 9.
- The cost of this merge is:

$$
5 + 9 = \mathbf{14}
$$

**Current Pool:**

$$
\{14\}
$$

Since only one file remains, we are done.

---

### Calculating Total Cost

To find the total work done, we add up the cost of every individual merge step:

$$
\text{Total Cost} = 5 + 9 + 14 = 28
$$

---

## 4️⃣ The "Proof": Why is this Optimal?

To prove to students that the Greedy choice matters, let's look at what happens if we merge them in the wrong order (largest first):

- **Bad Step 1:** Merge 5 and 4. Cost = 9. (Pool: 9, 3, 2)
- **Bad Step 2:** Merge 9 and 3. Cost = 12. (Pool: 12, 2)
- **Bad Step 3:** Merge 12 and 2. Cost = 14. (Pool: 14)

**Total Cost of Bad Strategy:**

$$
9 + 12 + 14 = \mathbf{35}
$$

By simply choosing the smallest files first, our Optimal Merge Pattern reduced the total operations from 35 down to 28. We saved the computer significant processing time!

# 4️⃣ Prim’s Algorithm (MST)

<div style="background-color:#FDEDEC; padding:15px; border-radius:10px;">

### 📌 Definition
Finds the Minimum Spanning Tree of a connected weighted graph.

### 🧠 Greedy Choice
Pick the smallest edge connected to the current tree.

</div>

---

### 🔢 Example Graph

Edges:

- A–B (2)  
- B–C (3)  
- A–C (5)  

Start from $A$.

---

### ✅ Solution

1. Pick A–B (2)  
2. Pick B–C (3)  

$$
\textbf{MST Weight} = 2 + 3 = 5
$$

---

# 5️⃣ Dijkstra’s Algorithm (SSSP)

<div style="background-color:#F4ECF7; padding:15px; border-radius:10px;">

### 📌 Definition
Find shortest path from a source vertex to all other vertices.

### 🧠 Greedy Choice
Pick the unvisited vertex with the smallest tentative distance.

</div>

---

### 🔢 Example

Source = $A$

Edges:

- $A \rightarrow B$ (10)  
- $A \rightarrow C$ (3)  
- $C \rightarrow B$ (2)  

---

### ✅ Steps

1. Initialize distances:

$$
A = 0, \quad B = \infty, \quad C = \infty
$$

2. Visit $A$ → Update $B = 10$, $C = 3$

3. Visit $C$ → New distance to $B$:

$$
3 + 2 = 5
$$

Since $5 < 10$, update $B = 5$.

---

### 🏆 Final Distances

| Vertex | Distance |
|--------|----------|
| A | 0 |
| C | 3 |
| B | 5 |

---

# 6️⃣ Kruskal’s Algorithm (MST)

<div style="background-color:#EAF2F8; padding:15px; border-radius:10px;">

### 📌 Definition
Finds MST by sorting edges globally.

### 🧠 Greedy Choice
Pick smallest edge. Skip if it forms a cycle.

</div>

---

### 🔢 Example

Edges:

- A–B (1)  
- B–C (2)  
- A–C (3)  

---

### ✅ Solution

1. Pick A–B (1)  
2. Pick B–C (2)  
3. Skip A–C (3) → Forms cycle  

$$
\textbf{MST Weight} = 1 + 2 = 3
$$

# 📊 Summary Table

| Algorithm | Greedy Choice | Goal |
|------------|--------------|------|
| Fractional Knapsack | Highest $v/w$ | Maximize Value |
| Huffman | Two smallest frequencies | Minimize Encoding Cost |
| Optimal Merge | Two smallest files | Minimize Merge Cost |
| Prim’s | Smallest edge from tree | Build MST |
| Dijkstra’s | Smallest distance vertex | Shortest Path |
| Kruskal’s | Smallest edge globally | Build MST |

# 🎓 Key Insight

Greedy algorithms work when:

1. **Greedy Choice Property** holds  
2. **Optimal Substructure** exists  

Local optimal decisions lead to global optimal solutions.
